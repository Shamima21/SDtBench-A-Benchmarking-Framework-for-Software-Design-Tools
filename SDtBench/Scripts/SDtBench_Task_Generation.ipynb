```python
# Generate DAC files (.puml, .mmd, .dot, .dsl, .d2) for the given E-commerce Checkout task
# and save them to /mnt/data. Also bundle them in a ZIP for convenience.

import os, textwrap, zipfile

out_dir = "/mnt/data/ecommerce_dac"
os.makedirs(out_dir, exist_ok=True)

entities = [
    "Customer", "Order", "Product", "Cart", "Invoice",
    "Payment", "Shipping", "Supplier", "Category", "Review"
]

# Relationships (tuples): (A, relation, B)
# Interpreted as A --> B (association/uses/dependsOn/etc. depending on the DSL)
rels = [
    ("Customer", "places", "Order"),
    ("Customer", "owns", "Cart"),
    ("Cart", "contains", "Product"),
    ("Order", "includes", "Product"),
    ("Order", "generates", "Invoice"),
    ("Order", "settledBy", "Payment"),
    ("Order", "deliveredVia", "Shipping"),
    ("Product", "providedBy", "Supplier"),
    ("Product", "categorizedAs", "Category"),
    ("Review", "about", "Product"),
    ("Customer", "writes", "Review"),
]

# ---------- 1) PlantUML (.puml) ----------
puml = ["@startuml", "skinparam classAttributeIconSize 0", "title E-commerce Checkout Class Diagram"]
for e in entities:
    puml.append(f"class {e}")
for a, rel, b in rels:
    puml.append(f"{a} --> {b} : {rel}")
puml.append("@enduml")

puml_path = os.path.join(out_dir, "ecommerce.puml")
with open(puml_path, "w") as f:
    f.write("\n".join(puml))

# ---------- 2) Mermaid (.mmd) ----------
mmd = ["---", "title: E-commerce Checkout Class Diagram", "---", "classDiagram"]
for e in entities:
    mmd.append(f"class {e}")
for a, rel, b in rels:
    # Mermaid classDiagram relation: A --> B : label
    mmd.append(f"{a} --> {b} : {rel}")
mmd_path = os.path.join(out_dir, "ecommerce.mmd")
with open(mmd_path, "w") as f:
    f.write("\n".join(mmd))

# ---------- 3) Graphviz (.dot) ----------
dot = [
    "digraph ECommerce {",
    '  rankdir=LR;',
    '  labelloc="t";',
    '  label="E-commerce Checkout Entity Graph";',
    '  node [shape=box, style=rounded];'
]
for e in entities:
    dot.append(f'  "{e}";')
for a, rel, b in rels:
    dot.append(f'  "{a}" -> "{b}" [label="{rel}"];')
dot.append("}")
dot_path = os.path.join(out_dir, "ecommerce.dot")
with open(dot_path, "w") as f:
    f.write("\n".join(dot))

# ---------- 4) Structurizr DSL (.dsl) ----------
# We'll model this as a Software System with containers for each entity, and relationships.
dsl = [
'workspace "E-commerce Checkout" "Modeled as entity containers" {',
"  model {",
'    softwareSystem ecommerce "E-commerce Platform" {'
]
for e in entities:
    dsl.append(f'      container {e.lower()} "{e}" "Entity" "Represents the {e} concept"')
# relationships
for a, rel, b in rels:
    dsl.append(f"      {a.lower()} -> {b.lower()} \"{rel}\"")
dsl += [
"    }",
"  }",
'  views {',
'    systemContext ecommerce "E-commerce Context" {',
"      include *",
'      autoLayout lr',
"    }",
'    container ecommerce "E-commerce Entities" {',
"      include *",
'      autoLayout lr',
"    }",
"    theme default",
"  }",
"}"
]
dsl_path = os.path.join(out_dir, "ecommerce.dsl")
with open(dsl_path, "w") as f:
    f.write("\n".join(dsl))

# ---------- 5) D2 (.d2) ----------
d2 = [
'# E-commerce Checkout Graph (D2)',
'direction: right',
'style: { shape: rounded }',
]
for e in entities:
    d2.append(f'{e}: "{e}"')
for a, rel, b in rels:
    d2.append(f'{a} -> {b}: "{rel}"')
d2_path = os.path.join(out_dir, "ecommerce.d2")
with open(d2_path, "w") as f:
    f.write("\n".join(d2))

# Zip everything
zip_path = "/content/sample_data/ecommerce_dac_files.zip"
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for path in [puml_path, mmd_path, dot_path, dsl_path, d2_path]:
        zf.write(path, arcname=os.path.basename(path))

(puml_path, mmd_path, dot_path, dsl_path, d2_path, zip_path)
```
